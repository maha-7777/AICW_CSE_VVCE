```python
from datetime import date 
```

```python
class student:
    def __init__(self,name,age):
        self.name = name
        self.age = age
        
    @classmethod
    def calculate_age(cls,name,birthyear):
        return cls(name,date.today().year - birthyear)

    def show(self):
        print(self.name + "'s age is'" + str(self.age) )
```

```python
age = student('MLK',19)
age.show()

```

```python
by = student.calculate_age('MLK',2006)
by.show()
```

```python
class school:
    # class variable
    name = "TMS"
    @classmethod
    def school_name(cls):
        print('school name is', cls.name)
    # create class method
    school.school_name=classmethod(school.school_name)

    # call class method
school.school_name()

    
```

```python
# static method
class emp:
    @staticmethod
    def sample(x):
        print('inside static method', x)
    # call static method
emp.sample(10)
# can be called using object
emp1=emp()
emp1.sample(10)
```

```python
class bankaccount:
    def __init__(self,name,balance=0):
        # instance attribute
        self.name = name
        self.balance = balance
    # deposit money
    def deposit(self,amount):
        if amount > 0:
            self.balance += amount
            print("deposit sucessfully!")
            print("deposited:", amount)
        else:
            print("invalid deposit amount")
        # withdraw money
    def withdraw(self,amount):
            if amount <=0:
                print('invalid withdrawl amount !')
            elif amount > self.balance:
                print('insufficent balance!')
            else:
                self.balance -= amount
                print("withdrawl succesfull!")
                print("withdrawn:", amount)
        #check balance
    def check_balance(self):
            print("current balance:",self.balance)
        #display account details
    def display_details(self):
            print("account holder:",self.name)
            print("balance:",self.balance)
```

```python
acc1=bankaccount("MLK",1000)
```

```python
acc1.display_details()
```

```python
acc1.deposit(5000)
```

```python
acc1.check_balance()
```

```python
acc1.withdraw(2000)
```

```python
acc1.check_balance()
```

# INHERITANCE, POLYMORPHISM AND ENCAPSULATION

```python
# parent class
class person:
    def person_info(self,name,age):
        print('inside person class')
        print('name:',name,'age:',age)

#parent class 2
class company:
    def company_info(self,company_name,location):
        print('inside the company class')
        print('name:',company_name,'location',location)
```

```python
# child class
class employee(person,company):
    def employee_info(self,salary,skill):
        print('inside employee class - child class')
        print('salary',salary,'skill:',skill)
```

```python
# create an object of empoyee
emp2 = employee()
```

```python
# access the  data
emp2.person_info('MLK',19)
emp2.company_info('edunet foundation','bengaluru')
emp2.employee_info(25000,'AIML engineer')
```

```python
class vehicle:
    def info(self):
        print('this is a vehicle')
class car(vehicle):
    def car_info(self,name):
        print('car name is:',name)
class truck(vehicle):
    def truck_info(self,name):
        print('truck name is:', name)
```

```python
obj1 = car()
obj1.info()
obj1.car_info('tata')
```
