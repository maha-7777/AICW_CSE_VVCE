```python
class student:
    def __init__(self,name,age):
        self.name = name
        self.age = age
        
    @classmethod
    def calculate_age(cls,name,birthyear):
        return cls(name,date.today().year - birthyear)

    def show(self):
        print(self.name + "'s age is'" + str(self.age) )
```

```python
age = student('MLK',19)
age.show()

```

```python
by = student.calculate_age('MLK',2006)
by.show()
```

```python
class bankaccount:
    def __init__(self,name,balance=0):
        # instance attribute
        self.name = name
        self.balance = balance
    # deposit money
    def deposit(self,amount):
        if amount > 0:
            self.balance += amount
            print("deposit sucessfully!")
            print("deposited:", amount)
        else:
            print("invalid deposit amount")
        # withdraw money
    def withdraw(self,amount):
            if amount <=0:
                print('invalid withdrawl amount !')
            elif amount > self.balance:
                print('insufficent balance!')
            else:
                self.balance -= amount
                print("withdrawl succesfull!")
                print("withdrawn:", amount)
        #check balance
    def check_balance(self):
            print("current balance:",self.balance)
        #display account details
    def display_details(self):
            print("account holder:",self.name)
            print("balance:",self.balance)
```

```python
acc1=bankaccount("MLK",1000)
```

```python
acc1.display_details()
```

```python
acc1.deposit(5000)
```

```python
acc1.check_balance()
```

```python
acc1.withdraw(2000)
```

```python
acc1.check_balance()
```

# INHERITANCE, POLYMORPHISM AND ENCAPSULATION

```python
# parent class
class person:
    def person_info(self,name,age):
        print('inside person class')
        print('name:',name,'age:',age)

#parent class 2
class company:
    def company_info(self,company_name,location):
        print('inside the company class')
        print('name:',company_name,'location',location)
```

```python
# child class
class employee(person,company):
    def employee_info(self,salary,skill):
        print('inside employee class - child class')
        print('salary',salary,'skill:',skill)
```

```python
# create an object of empoyee
emp2 = employee()
```

```python
# access the  data
emp2.person_info('MLK',19)
emp2.company_info('edunet foundation','bengaluru')
emp2.employee_info(25000,'AIML engineer')
```

```python
class vehicle:
    def info(self):
        print('this is a vehicle')
class car(vehicle):
    def car_info(self,name):
        print('car name is:',name)
class truck(vehicle):
    def truck_info(self,name):
        print('truck name is:', name)
```

```python
obj1 = car()
obj1.info()
obj1.car_info('tata')
```

```python
obj2 = truck()
obj2.info()
obj2.truck_info('eicher')
```

```python
# hybrid inheritence
class vehicle:
    def vehicle_info(self):
        print('inside vehicle class')
class car(vehicle):
    def car_info(self):
        print('inside car class')
class truck_info(vehicle):
    def truck_info(self):
        print('inside truck class')
class sportscar(car,vehicle):
    def sports_car_info(self):
        print('inside sportscar class')
```

```python
# create an object
s_car = sportscar()
s_car.vehicle_info()
s_car.car_info()
s_car.sports_car_info()
```

```python
# super() - function
class company:
    def company_name(self):
        return ' eduenet foundation '
class employee(company):
    def info(self):
        # call the super func
        c_name = super().company_name()
        print('MLK works at',c_name)
```

```python
emp3=employee()
emp3.info()
```

```python
# method overriding
class vehicle:
    def max_speed(self):
        print("max speed is 100 kmph")
class car(vehicle):
    # overriding the implementation of vehicle class
    def max_speed(self):
        print('max speed id 200 kmph')
```

```python
car = car()
car.max_speed()
```

```python
# public member
class employee:
    def __init__(self,name,salary):
        self.name=name
        self.salary=salary
    # public instance method
    def show(self):
        # access public data members
        print('name',self.name,'salary',self.salary)
```

```python
emp = employee('MLK',10000)
# access the public data members
print('name',emp.name,'salary',emp.salary)
emp.show()
```

```python
# private member
class employee:
    def __init__(self,name,salary):
        self.name = name # public  data member
        self.__salary = salary # private data member
```

```python
# create an object
emp = employee('MLK',5000)
# acces the private data member
print('salary',emp.__salary)
```

```python
# protected member
class company:
    def __init__(self):
        # protected member
        self._project = "NLP"
# child class
class employee(company):
    def __init__(self,name):
        self.name=name
        company.__init__(self)
    def show(self):
        print("employee name:",self.name)
        # access the protected member
        print('working on project:',self._project)
```

```python
c = employee("MLK")
c.show()
```

```python
# direct access
print('project:', c._project)
```

```python
class car:
    def __init__(self,brand,model):
        self.brand = brand
        self.model = model
    def move(self):
        print("drive!")
class boat:
    def __init__(self,brand,model):
        self.brand = brand
        self.model = model
    def move(self):
        print("sail!")
class plain:
    def __init__(self,brand,model):
        self.brand = brand
        self.model = model
    def move(self):
        print("fly!")
```

```python
# polymorphism with inheritance
class bird:
    def intro(self): 
        print('there are many types of birds!')
    def flight(self):
        print('most of the birds can fly but some cant!')
class sparrow(bird):
    def flight(self):
        print("sparrows can fly!")
class ostrich(bird):
    def flight(self):
        print("ostrich cant fly!")
        
```

```python
ob_bird = bird()
ob_bird.intro()
ob_bird.flight()
```

```python
ob_sparrow = sparrow()
ob_sparrow.flight()
```

```python
ob_ostrich = ostrich()
ob_ostrich.flight()
```

```python
# over loading
class complex:
    def __init__(self,a,b):
        self.a=a
        self.b=b
    def __add__(self,other):
        return self.a + other.a, self.b + other.b
ob1 = complex(1,2)
ob2 = complex(2,3)
ob3 = ob1+ob2
print(ob3)
```
