```python
from datetime import date 
```

```python
class student:
    def __init__(self,name,age):
        self.name = name
        self.age = age
        
    @classmethod
    def calculate_age(cls,name,birthyear):
        return cls(name,date.today().year - birthyear)

    def show(self):
        print(self.name + "'s age is'" + str(self.age) )
```

```python
age = student('MLK',19)
age.show()

```

```python
by = student.calculate_age('MLK',2006)
by.show()
```

```python
class school:
    # class variable
    name = "TMS"
    @classmethod
    def school_name(cls):
        print('school name is', cls.name)
    # create class method
    school.school_name=classmethod(school.school_name)

    # call class method
school.school_name()

    
```

```python
# static method
class emp:
    @staticmethod
    def sample(x):
        print('inside static method', x)
    # call static method
emp.sample(10)
# can be called using object
emp1=emp()
emp1.sample(10)
```
