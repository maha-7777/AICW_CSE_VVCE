```python
class student:
    def __init__(self,usn,name,age):
        # inastance variable
        self.usn=usn
        self.name=name
        self.age=age
    # instance method access instance variable
    def show(self):
        print('USN',self.usn,'Name',self.name, 'Age',self.age)
    # instance method to modify instance variable
    def update(self,usn,age):
        self.usn=usn
        self.age=age
```

```python
# create an object
stu = student(156, 'MLK' ,19)
stu.show()
```

```python
stu.update(236,12)
stu.show()
```
